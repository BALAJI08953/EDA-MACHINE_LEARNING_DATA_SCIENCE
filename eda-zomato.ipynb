```python
df_country=pd.read_excel('Country-Code.xlsx')
df_country.head()
```

```python
df.columns
```

```python
final_df=pd.merge(df,df_country,on='Country Code',how='left')
```

```python
final_df.head()
```

```python
final_df.columns
```

```python
country_names=final_df.Country.value_counts().index
country_names
```

```python
cvalues=final_df.Country.value_counts().values
```

```python
plt.pie(cvalues,labels=country_names)
```

```python
#pie chart for top 3 companies which uses zomato
plt.pie(cvalues[:3],labels=country_names[:3],autopct='%.2f%%')
```

```python
#observations:zomato maximum records or transaction are from india
#after that usa and then united kingdoms
```

```python
#numerical variables
ratings=final_df.groupby(['Aggregate rating','Rating color','Rating text']).size().reset_index().rename(columns={0:'Rating_count'})
```

```python
ratings
```

```python
##Observation
#where ratings is between 4.5 to 4.9 ----> excellent
#from 4.0 to 4.4 is between  ---> very good 
#when rating is between 3.5 to 3.9 -----> good
#when rating is between 3.0 to 3.4 ----> average
#when the rating is between 2.5 to 2.9 ---->average
#when the rating is between 2.0 to 2.4 ---> poor
```

```python
#diagrams 
ratings.head()
```

```python
import matplotlib
matplotlib.rcParams['figure.figsize']=(15,10)
sns.barplot(x='Aggregate rating',y='Rating_count',hue='Rating color',data=ratings,palette=['white','red','orange','yellow','green','green'])
```

```python
#onservations 
#not rated count is very high 
#maxximum numbers of rating are between are between 2.4 to 3.4
```

```python
##count plot 
sns.countplot(x='Rating color',data=ratings,palette=['white','red','orange','yellow','green','green'])
```
